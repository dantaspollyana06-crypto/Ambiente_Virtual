```python
import pandas as pd
#df=dataframe
dados_texto= pd.read_csv("citacoes.csv")
print("._"*50)
dados_numericos = pd.read_csv("analise_numpy.csv")
df = pd.merge(
    dados_texto,
    dados_numericos,
    on="ID"
)
print(df["Autor"].unique())
# A função unique() retorna um array com os valores únicos da coluna "Autor". Isso significa que ele vai listar todos os autores diferentes que aparecem no dataframe, sem repetir nenhum deles.
print(".-"*50)
print(f"---> São {df["Autor"].nunique()} autores")
# A função nunique() retorna o número de valores únicos da coluna "Autor". Isso significa que ele vai contar quantos autores diferentes aparecem no dataframe, sem contar nenhum deles mais de uma vez.
df.rename(columns={"Quantidade_Tags": "Total_Tags"}, inplace=True)
df["Média_Caracteres_Tags"]=(df["Caracteres"]/df["Total_Tags"])
df
# A função rename() é usada para renomear as colunas do dataframe. O parâmetro columns é um dicionário onde a chave é o nome antigo da coluna e o valor é o novo nome da coluna. O parâmetro inplace=True significa que a alteração será feita no próprio dataframe, sem precisar criar uma cópia.
```

```python
import pandas as pd
#df=dataframe
dados_texto= pd.read_csv("citacoes.csv")
print("._"*50)
dados_numericos = pd.read_csv("analise_numpy.csv")
df = pd.merge(
    dados_texto,
    dados_numericos,
    on="ID" )

grupo = df.groupby("Autor")["Citação"].count()
print(grupo)
# A função groupby() agrupa o dataframe pelo valor da coluna "Autor" e, em seguida, a função count() conta o número de citações para cada autor. O resultado é um novo dataframe com os autores como índice e o número de citações como valores.

print("._"*50)
grupo2 = df.groupby("Autor")["Caracteres"].mean()
print(grupo2)

 #A função groupby() agrupa o dataframe pelo valor da coluna "Autor" e, em seguida, a função mean() calcula a média do número de caracteres para cada autor. O resultado é um novo dataframe com os autores como índice e a média de caracteres como valores.

df.drop(columns=["Média_Caracteres_Tags"], inplace=True)
df
# A função drop() é usada para remover colunas do dataframe. O parâmetro columns é uma lista com os nomes das colunas a serem removidas. O parâmetro inplace=True significa que a alteração será feita no próprio dataframe, sem precisar criar uma cópia.
df["Valor_teste"]=None
df.fillna(0)
```
