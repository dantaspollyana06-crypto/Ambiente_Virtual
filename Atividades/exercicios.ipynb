```python
import numpy as np
"""1 – Crie
dois arrays NumPy: A = [10, 20, 30, 40, 50] e B = [1, 2, 3, 4, 5].  Realize as seguintes operações entre
eles:     

Soma
Subtração
Multiplicação
Divisão """

a = np.array([10,20,30,40,50])
b = np.array([1,2,3,4,5])
c = np.add(a,b)
d = np.subtract(a,b)
e = np.multiply(a,b)
f = np.divide(a,b)
print(f" A Soma é:---> {c}\n A Subtração é:--->{d}\n A Multiplicação é: --->{e}\n A divisão é:---> {f}")
```
