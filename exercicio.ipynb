```python
import pandas as pd
#df=dataframe
dados_texto= pd.read_csv("citacoes_final.csv")
dados_numericos = pd.read_csv("analise_numpy.csv")
df = pd.merge(
    dados_texto,
    dados_numericos,
    on=["ID", "Caracteres", "Quantidade_Tags"]
)

longa=df[df["Caracteres"]>100]
print("↪ As citações com mais de 100 caracteres são:")
print(longa[["Autor","Citação","Caracteres"]])
print("¤"*70)
#
especifico=df[df["Autor"]=="Albert Einstein"]
print("↪ As citações do autor Albert Einstein são:")        
print(especifico[["Autor","Citação"]])
print("¤"*70)
tags=df[df["Quantidade_Tags"]<=4]
print("↪ As citações com 4 ou menos tags são:")
print(tags[["Autor","Citação","Quantidade_Tags"]])
print("¤"*70)
frases=df[(df['Caracteres']>=100) & (df["Quantidade_Tags"]>=3)]
print("↪ As citações com mais de 100 caracteres e pelo menos 3 tags são:")
print(frases[["Citação","Caracteres","Quantidade_Tags"]])
print("¤"*70)
```

```python

import pandas as pd
#df=dataframe
dados_texto= pd.read_csv("citacoes_final.csv")
dados_numericos = pd.read_csv("analise_numpy.csv")
df = pd.merge(
    dados_texto,
    dados_numericos,
    on=["ID", "Caracteres", "Quantidade_Tags"]
)

frases_ordenadas = df.sort_values(by="Caracteres", ascending=False)
print("↪ As citações ordenadas pela quantidade de caracteres (da maior para a menor) são:")
print(frases_ordenadas[["Citação","Caracteres"]])
print("¤"*70)
maior_citacao = df.loc[df["Caracteres"].idxmax()]
print("↪A maior citação é:")
print(maior_citacao[["Citação","Caracteres","Autor"]])
print("¤"*70)
print("↪A quantidade de citações por autor é:")
print(df["Autor"].value_counts())
print("¤"*70)
print("↪ O número de autores diferentes no dataset é:")
print(df["Autor"].nunique())
```
